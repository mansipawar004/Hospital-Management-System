```python
# Hospital Management System:
# parent class
class Person:
  def __init__(self,id,name,age):
    self.id=id
    self.name=name
    self.age=age
  def display(self):
    print("Id:",self.id)
    print("Name:",self.name)
    print("Age:",self.age)

#Inheritance
class Doctor(Person):
   def __init__(self, id, name, age, specialization):
        super().__init__(id, name, age)
        self.specialization = specialization
   def display(self):
    super().display()
    print("specialization:",self.specialization)

class Patient(Person):
   def __init__(self, id, name, age, disease):
        super().__init__(id, name, age)
        self.disease = disease
   def display(self):
    super().display()
    print("Disease:",self.disease)

# Appointment
class Appointment:
  def __init__(self,patient_name,doctor_name,date):
    self.patient_name=patient_name
    self.doctor_name=doctor_name
    self.date=date
  def show (self):
    print("      Appointment Details     ")
    print("Patient:",self.patient_name)
    print("Doctor:",self.doctor_name)
    print("Date:",self.date)

# Encapsulation
class Billing:
  def __init__(self,amount):
    self.__amount=amount
  def get_bill(self):
    return self.__amount
  def display(self):
    print("   Bill   ")
    print("Total Bill:",self.__amount)

#Abstraction
from abc import ABC,abstractmethod
class Payment(ABC):
  @abstractmethod
  def pay(self):
    pass

#Polymorphism
class CashPayment(Payment):
  def pay(self):
    print("payment done using cash")
class UPIPayment(Payment):
  def pay(self):
    print("payment done using UPI")
class CardPayment(Payment):
  def pay(self):
    print("payment done using debit/credit card")

# Pharmacy Module
class Pharmacy:
  def __init__(self,medicine,price):
    self.medicine=medicine
    self.price=price
  def display(self):
    print("  Pharmacy  ")
    print("Medicine:",self.medicine)
    print("Price:",self.price)

#Laboratory Module
class Laboratory:
  def __init__(self,test_name):
    self.test_name=test_name
  def display(self):
    print("   Laboratory  ")
    print("Test:",self.test_name)

# Feedback Module
class Feedback:
  def __init__(self,message):
    self.message=message
  def display(self):
    print("  Feedback  ")
    print("Feedback:",self.message)

# Main Menu
doctor=None
patient=None
appointment = None
bill = None
pharmacy = None
lab = None
feedback = None
payment = None

username = input("Enter Username : ")
password = input("Enter Password : ")

if username == "admin" and password == "123":
    print("Login Successful")
else:
    print("Invalid Username or Password")
    exit()

while True:
  print("   HOSPITAL MANAGEMENT SYSTEM   ")
  print("1.Add Doctor")
  print("2.Add Patient")
  print("3.View Doctor")
  print("4.View Patient")
  print("5.Book Appointment")
  print("6.Generate Bill")
  print("7. Payment")
  print("8. Pharmacy")
  print("9. Laboratory")
  print("10. Feedback")
  print("11. Exit")

  choice=int(input("Enter your choice"))

  if choice==1:
    did=int(input("Enter Doctor id:"))
    dname=(input("Enter Doctor name:"))
    dage=int(input("Enter age:"))
    dspecialization=(input("Enter specialization:"))

    doctor=Doctor(did,dname,dage,dspecialization)
    print("Doctor Added Successfully")

  elif choice ==2:
    pid=int(input("Enter Patient id:"))
    pname=(input("Enter Patient name:"))
    page=int(input("Enter  age:"))
    pdisease=(input("Enter Disease:"))

    patient=Patient(pid,pname,page,pdisease)
    print("Patient Added Successfully")

  elif choice ==3:
    if doctor:
      doctor.display()
    else:
      print("No Doctor Record Found")

  elif choice ==4:
    if patient:
      patient.display()
    else:
      print("No Patient Record Found")

  elif choice==5:
    if patient and doctor:
      date=(input("Enter Appointment Date:"))
      appointment = Appointment( patient.name, doctor.name, date)
      appointment.show()
    else:
        print("Please Add Doctor and Patient First")

  elif choice ==6:
    amount=int(input("Enter Bill Amount:"))
    bill=Billing(amount)
    bill.display()

  elif choice == 7:
    print("1. Cash")
    print("2. UPI")
    print("3. Card")
    pay = int(input("Choose Payment Method : "))

    if pay == 1:
        payment = CashPayment()

    elif pay == 2:
        payment = UPIPayment()

    elif pay == 3:
        payment = CardPayment()

    else:
        print("Invalid Choice")
        continue

    payment.pay()

  elif choice == 8:
    medicine = input("Enter Medicine Name : ")
    price = int(input("Enter Price : "))
    ph = Pharmacy(medicine, price)
    ph.display()

  elif choice == 9:
    test = input("Enter Test Name : ")
    lab = Laboratory(test)
    lab.display()

  elif choice == 10:
    msg = input("Enter Feedback : ")
    fb = Feedback(msg)
    fb.display()



  elif choice == 11:
    print("Thank You")
    break

  else:
    print("Invalid Choice")
```
